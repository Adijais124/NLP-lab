Q1

```python
import nltk
from nltk.corpus import brown, stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
import string

nltk.download('brown')
nltk.download('punkt_tab')
nltk.download('stopwords')

sentences = brown.sents(categories='news')[:20]

stop_words = set(stopwords.words('english'))
punct = set(string.punctuation)

for i, sent in enumerate(sentences[:5]):
    original = " ".join(sent)
    sentence_tokens = sent_tokenize(original)
    word_tokens = [w.lower() for w in word_tokenize(original)]
    cleaned_tokens = [w for w in word_tokens if w.isalpha() and w not in stop_words]

    print(f"\nSentence {i+1}: {original}")
    print("Sentence Tokens:", sentence_tokens)
    print("Word Tokens:", word_tokens)
    print("Cleaned Tokens:", cleaned_tokens)
```

Q2

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
import pandas as pd

nltk.download('wordnet')

words = ['playing', 'studies', 'running', 'better', 'cars'] + [w for w in brown.words(categories='news')[:25] if w.isalpha()]

ps = PorterStemmer()
lemmatizer = WordNetLemmatizer()

data = {
    'Original': words,
    'Stemmed': [ps.stem(w) for w in words],
    'Lemmatized': [lemmatizer.lemmatize(w) for w in words]
}

df = pd.DataFrame(data)
print(df.head(15))
```

Q3

```python
from nltk.corpus import words
from difflib import get_close_matches

nltk.download('words')

word_list = set(words.words())
misspelled = ["speling", "studing", "runing", "bettar", "carz", "machin", "lernng", "nlp", "artifical", "inteligence"]

for w in misspelled:
    suggestions = get_close_matches(w, word_list, n=3, cutoff=0.8)
    print(f"{w} -> Suggestions: {suggestions}")

sentences = [
    "I am studing NLP.",
    "Machin lernng is fun.",
    "Cars are runing fast.",
    "Artifical inteligence is powerful.",
    "He is bettar than me."
]

corrected = []
for s in sentences:
    corrected_sent = []
    for word in word_tokenize(s):
        if word.lower() not in word_list:
            suggestion = get_close_matches(word.lower(), word_list, n=1)
            corrected_sent.append(suggestion[0] if suggestion else word)
        else:
            corrected_sent.append(word)
    corrected.append(" ".join(corrected_sent))

print("\nCorrected Sentences:")
for c in corrected:
    print(c)
```

```python
from nltk import FreqDist, ngrams

tokens = [w.lower() for w in brown.words(categories='news') if w.isalpha()]

unigrams = FreqDist(tokens)
print("Top 10 Unigrams:", unigrams.most_common(10))

bigrams = FreqDist(ngrams(tokens, 2))
print("Top 10 Bigrams:", bigrams.most_common(10))

trigrams = FreqDist(ngrams(tokens, 3))
print("Top 10 Trigrams:", trigrams.most_common(10))
```
