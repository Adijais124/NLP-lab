**Lowercasing**

```python


lowercase_paragraphs = []

for paragraph in paragraphs:
    lowercase_paragraphs.append(paragraph.lower())

# Display the first paragraph after lowercasing
print(lowercase_paragraphs[4])
```

**Whitespaces**

```python
# A3: Remove extra whitespace

cleaned_whitespace = []

for paragraph in lowercase_paragraphs:
    paragraph = " ".join(paragraph.split())
    cleaned_whitespace.append(paragraph)

# Display the first paragraph
print(cleaned_whitespace[0])
```

**contractions**

```python


contractions = {
    "can't": "cannot",
    "it's": "it is",
    "i'm": "i am",
    "i've": "i have",
    "i'll": "i will",
    "i’d": "i would",
    "i’m": "i am",
    "i’ve": "i have",
    "u're": "you are",
    "c'mon": "come on",
    "i <3 nlp": "i love nlp",
    "i <3 tech": "i love tech"
}

contraction_cleaned = []

for paragraph in cleaned_whitespace:
    for contraction, replacement in contractions.items():
        paragraph = paragraph.replace(contraction, replacement)

    contraction_cleaned.append(paragraph)


print(contraction_cleaned[0])
```

Remove special **character**

```python

special_char_cleaned = []

for paragraph in contraction_cleaned:
    # Remove special characters except URL-related characters
    paragraph = re.sub(r'[^a-zA-Z0-9\s@#:/._-]', '', paragraph)
    special_char_cleaned.append(paragraph)

print(special_char_cleaned[0])
```

**Reduce duplicate letters**

```python


duplicate_cleaned = []

for paragraph in special_char_cleaned:

    paragraph = re.sub(r'(.)\1{2,}', r'\1\1', paragraph)
    duplicate_cleaned.append(paragraph)

print(duplicate_cleaned[0])
```

** punctuation**

```python


punctuation_cleaned = []

for paragraph in duplicate_cleaned:
    # Replace repeated punctuation marks with a single punctuation mark
    paragraph = re.sub(r'([.!?,])\1+', r'\1', paragraph)

    punctuation_cleaned.append(paragraph)

print(punctuation_cleaned[0])
```

** Remove URLs**

```python


url_cleaned = []

for paragraph in punctuation_cleaned:

    paragraph = re.sub(r'https?://\S+|www\.\S+', '', paragraph)

    url_cleaned.append(paragraph)

print(url_cleaned[0])
```

**Final whitespace cleanup**

```python


final_cleaned_paragraphs = []

for paragraph in url_cleaned:
    # Replace multiple spaces with a single space
    paragraph = " ".join(paragraph.split())

    final_cleaned_paragraphs.append(paragraph)

# Display all cleaned paragraphs

for i, paragraph in enumerate(final_cleaned_paragraphs, start=1):
    print(f"\nParagraph {i}:")
    print(paragraph)
```
